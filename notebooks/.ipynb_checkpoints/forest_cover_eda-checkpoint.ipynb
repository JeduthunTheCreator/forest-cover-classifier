Exploratory Data Analysis 

Load and inspect data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# load the dataset to a dataframe
df = pd.read_csv('cover_data.csv')

# Basic info
print(df.info())
print(df.describe())
```

Class distribution

```python
print(df['class'].value_counts())
sns.countplot(x='class', data=df)
```

Missing values

```python
print(df.isnull().sum())
```

Continuous feature distributions

```python
# 4. Correlation heatmap for continuous features
continuous_features = df.iloc[:, :10]  # first 10 are continuous
sns.heatmap(continuous_features.corr(), annot=True, cmap='coolwarm')
```

```python
# 5. Feature distributions
fig, axes = plt.subplots(2, 5, figsize=(18, 16))
continuous_features.hist(
    ax=axes,
    bins=30,             # more bins for smoother distribution view
    color='steelblue',
    edgecolor='black',   # bar borders for clarity
    linewidth=0.5
)

plt.suptitle('Continuous Feature Distributions', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout(pad=3.0)          # increase padding between subplots
plt.show()
```

```python
# print the number of rows and columns
print(continuous_features.shape)
```

Skewed features

```python
# identify skewed features 
print(continuous_features.skew().sort_values(ascending=False))
```

Wilderness area distribution

```python
# Wilderness areas (4 columns)
wilderness_cols = [col for col in df.columns if 'Wilderness' in col]
df[wilderness_cols].sum().plot(kind='bar', figsize=(8, 4), color='steelblue', edgecolor='black')
plt.title('Wilderness Area Distribution')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Check if class imbalance correlates with Wilderness Area 1
print(df.groupby('class')['Wilderness_Area1'].sum().sort_values(ascending=False))
```

Soil type distribution

```python
# Soil types (40 columns)
soil_cols = [col for col in df.columns if 'Soil' in col]
df[soil_cols].sum().plot(kind='bar', figsize=(16, 4), color='steelblue', edgecolor='black')
plt.title('Soil Type Distribution')
plt.ylabel('Count')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# identify low frequency soil types
soil_counts = df[soil_cols].sum().sort_values()
print(soil_counts[soil_counts < 1000])   # soil types with fewer than 1000 samples
```

Feature vs Target Analysis 

```python
# Box plot of each continuous feature grouped by cover type
fig, axes = plt.subplots(2, 5, figsize=(20, 8))
for idx, col in enumerate(continuous_features.columns):
    ax = axes[idx//5, idx%5]
    df.boxplot(column=col, by='class', ax=ax)
    ax.set_title(col)
    ax.set_xlabel('Cover Type')
plt.suptitle('Feature Distribution by Cover Type')
plt.tight_layout()
plt.show()
```

Basic Outlier Inspection

```python
# IQR based outlier count per feature
Q1 = continuous_features.quantile(0.25)
Q3 = continuous_features.quantile(0.75)
IQR = Q3 - Q1
outliers = ((continuous_features < (Q1 - 1.5 * IQR)) | 
            (continuous_features > (Q3 + 1.5 * IQR))).sum()
print(outliers.sort_values(ascending=False))
```
